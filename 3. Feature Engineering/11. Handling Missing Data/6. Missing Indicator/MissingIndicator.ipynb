```python

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split

from sklearn.impute import MissingIndicator, SimpleImputer
```

```python
df = pd.read_csv('train.csv',usecols=['Age','Fare','Survived'])
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python

X = df.drop(columns=['Survived'])
y = df['Survived']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
X_train.head()
```

```python
X_test.head()
```

```python
si = SimpleImputer()
X_train_trf = si.fit_transform(X_train)
X_test_trf = si.transform(X_test)
```

```python
X_train_trf
```

```python
X_test_trf
```

```python
from sklearn.linear_model import LogisticRegression

clf = LogisticRegression()

clf.fit(X_train_trf,y_train)

y_pred = clf.predict(X_test_trf)

from sklearn.metrics import accuracy_score
accuracy_score(y_test, y_pred)
```

```python
# this is the accuracy before applying missingIndicator
```

```python
# Now using MissingIndicator
```

```python
mi = MissingIndicator()

mi.fit(X_train)
```

```python
mi.features_ 
```

```python
# means -> missing values is in 0th index col
```

```python
X_train_missing = mi.transform(X_train)
```

```python
X_train_missing  # row with NaN -> True, and without NaN -> False
```

```python
X_test_missing = mi.transform(X_test)
```

```python
X_test_missing
```

```python
# Crete new col Age_NA 

X_train['Age_NA'] = X_train_missing 
```

```python
X_train
```

```python
X_test['Age_NA'] = X_test_missing
```

```python
X_test
```

```python
si = SimpleImputer()
X_train_trf2 = si.fit_transform(X_train)
X_test_trf2 = si.transform(X_test)
```

```python

from sklearn.linear_model import LogisticRegression

clf = LogisticRegression()

clf.fit(X_train_trf2,y_train)

y_pred = clf.predict(X_test_trf2)

from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

```python
# WITH missingIndicator it is giving little high accuracy.
```

```python
# SimpleImputer has an attribute add_indicator=True (by default -> False)
```

```python
si = SimpleImputer(add_indicator=True)
```

```python
X_train = si.fit_transform(X_train)
X_test = si.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression

clf = LogisticRegression()

clf.fit(X_train_trf2,y_train)

y_pred = clf.predict(X_test_trf2)


from sklearn.metrics import accuracy_score

accuracy_score(y_test,y_pred)
```
