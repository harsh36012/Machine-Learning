### Working With Dates

```python
# Converting to datetime datetype

date['date'] = pd.to_datetime(date['date'])
```

```python
date.info()
```

### 1. Extract year

```python
date['date_year'] = date['date'].dt.year

date.sample(5)
```

### 2. Extract Month

```python
# dt.month -> month number

date['date_month_no'] = date['date'].dt.month

date.sample(5)
```

```python
# dt.month_name()

date['date_month_name'] = date['date'].dt.month_name()

date.sample(5)
```

### 3. Extract Days

```python
# date(2019-05-15) -> dt.day -> 15

date['date_day'] = date['date'].dt.day

date.sample(5)
```

```python
# day of week
# dt.dayofweek -> gives which day of a week
# monday -> 0  &  sunday -> 6 

date['date_dow'] = date['date'].dt.dayofweek

date.sample(5)
```

```python
# day of week - name  
# dt.day_name()

date['date_dow_name'] = date['date'].dt.day_name()

date.drop(columns=['product_id','city_id','orders']).head()
```

```python
# is weekend?

date['date_is_weekend'] = np.where(date['date_dow_name'].isin(['Saturday','Sunday']),1,0)

date.sample(5)
```

### 4. Extract Week of year

```python
# dt.isocalendar().week -> gives which week of a year is it. (total -> 52/53 weeks a year)

date['date_week_of_year'] = date['date'].dt.isocalendar().week

date.sample(5)
```

### 5. Extract Quarter

```python
# dt.quarter  (4 quarter a year)

date['quarter'] = date['date'].dt.quarter

date.sample(5)
```

### 6. Extract Semester

```python
date['semester'] = np.where(date['quarter'].isin([1,2]),1,2)

date.sample(5)
```

### 7. Extract Time elapsed between dates

```python
import datetime as dt

today = dt.datetime.today()

today
```

```python
today - date['date']
```

```python
(today - date['date']).dt.days
```

```python
# minute passed

np.round((today - date['date']) / np.timedelta64(1,"m"),0)
```

```python
# week passed

np.round((today - date['date']) / np.timedelta64(1,"W"),0)
```

### Working With Time

```python
time['hour'] = time['date'].dt.hour
time['min'] = time['date'].dt.minute
time['sec'] = time['date'].dt.second

time.sample(5)
```
