```python
import pandas as pd
import seaborn as sns
```

```python
tips = sns.load_dataset("tips")
```

```python
flights = sns.load_dataset('flights')
```

```python
iris = sns.load_dataset("iris")
```

```python
titanic = sns.load_dataset('titanic')
```

## EDA using Bivariate and Multivariate

### 1. Scatterplot (Numerical - Numerical)

```python
tips.sample(5)
```

```python
sns.scatterplot(x= tips['total_bill'], y= tips['tip'])
```

```python
sns.scatterplot(x=tips['total_bill'],y= tips['tip'], hue = tips['sex'])
```

```python
sns.scatterplot(x = tips["total_bill"],y = tips["tip"], hue=tips["sex"], style=tips["smoker"], size=tips["size"])
```

### 2. Bar plot (Numerical - Categorical)

```python
titanic.sample(5)
```

```python
sns.barplot(x= titanic['pclass'],y = titanic['survived'])
```

```python
sns.barplot(x= titanic['pclass'],y=titanic['survived'],hue = titanic['sex'])
```

```python
sns.barplot(x = titanic["pclass"],y = titanic["fare"], hue= titanic["sex"])
```

### 3. Box plot (Numerical - Categorical)

```python
sns.boxplot(x=titanic['sex'],y=titanic['age'],hue = titanic['survived'])
```

### 4. Distplot (Numerical - Categorical)

```python
sns.displot(x='age',data= titanic, kind = 'kde',hue=titanic['survived'])
```

### 5. HeatMap (Categorical - Categorical)

```python
titanic.sample(5)
```

```python
pd.crosstab(titanic['pclass'],titanic['survived'])
```

```python
sns.heatmap(pd.crosstab(titanic['pclass'],titanic['survived']))
# pclass 3 , survived 0 -> lightest -> means those are in this category are highest 
# who travel in pclass 3 -> most lost thier lives.
# those are in pclass 1 -> most are survived 
```

```python
flights
```

```python
flights.pivot_table(values = 'passengers', index = 'month', columns = 'year',observed=False)
```

```python
sns.heatmap(flights.pivot_table(values="passengers",index="month",columns="year", observed=False))
```

```python
sns.clustermap(flights.pivot_table(values="passengers", index="month", columns="year", observed=False))
```

```python
# to get percentage , that how much in each Pclass get survived
print(titanic.groupby("pclass")["survived"].mean()*100)
```

```python
(titanic.groupby("pclass")["survived"].mean()*100).plot(kind='bar')
```

```python
titanic
```

```python
titanic.groupby("embarked")["survived"].mean()*100
```

```python
(titanic.groupby("embarked")["survived"].mean()*100).plot(kind='bar')
```

```python
titanic.groupby('sex')['survived'].mean()*100
```

```python
(titanic.groupby('sex')['survived'].mean()*100).plot(kind='bar')
```
